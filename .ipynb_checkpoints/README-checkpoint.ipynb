
# Credit Card Approval Classification Using Machine Learning Classification Models

## Problem Statement

The objective of this project is to build and compare multiple machine learning classification models for Credit Card Approval Classification. The models are trained on a credit card fillup dataset and evaluated using various performance metrics such as Accuracy, AUC Score, Precision, Recall, F1 Score, and Matthews Correlation Coefficient (MCC).

A Streamlit web application is developed to allow users to upload test data, select a machine learning model, and view model evaluation results interactively.

---

## Dataset Description

### Dataset Name
Credit Card Filing details Dataset

### Source
Kaggle

### Dataset Characteristics

- Total Records: 1549
- Total Features: 20
- Target Variable: label
- Class 0 = Application Approved
- Class 1 = Application Rejected

### Features

The dataset contains:
```text
-Ind_ID: Client ID
-Gender: Gender information
-Car_owner: Having car or not
-Propert_owner: Having property or not
-Children: Count of children
-Annual_income: Annual income
-Type_Income: Income type
-Education: Education level
-Marital_status: Marital_status
-Housing_type: Living style
-Birthday_count: Use backward count from current day (0), -1 means yesterday.
-Employed_days: Start date of employment. Use backward count from current day (0). Positive value means, individual is currently unemployed.
-Mobile_phone: Any mobile phone
-Work_phone: Any work phone
-Phone: Any phone number
-EMAIL_ID: Any email ID
-Type_Occupation: Occupation
-Family_Members: Family size
-Another data set (Credit_card_label.csv) contains two key pieces of information
-ID: The joining key between application data and credit status data, same is Ind_ID
-Label: 0 is application approved and 1 is application rejected
```

The dataset satisfies the assignment requirements:

- Number of Instances > 500
- Number of Features > 12

## Observation on result

| Model | Observation |
|---------|-------------|
| **Logistic Regression** | Achieved **89.27% accuracy**, but failed to identify any positive class instances, resulting in **Precision, Recall, F1-Score, and MCC of 0**. Although the **AUC Score (0.679)** indicates moderate discrimination ability, the model did not effectively handle the minority class. High accuracy is misleading due to class imbalance. |
| **Decision Tree** | Achieved **87.32% accuracy** with a balanced performance across minority class metrics. It recorded **43.33% Precision**, **59.09% Recall**, **50% F1-Score**, and **0.436 MCC**, indicating a reasonable ability to detect positive cases. Suitable as an interpretable baseline model. |
| **K-Nearest Neighbors (KNN)** | Achieved **88.78% accuracy** and **0.709 AUC Score**, but suffered from very low **Recall (9.09%)**, meaning most positive cases were missed. The low **F1-Score (14.81%)** and **MCC (0.150)** indicate limited effectiveness for detecting the minority class. |
| **Naïve Bayes** | Produced **89.27% accuracy**, but failed to classify any positive instances, resulting in **Precision, Recall, F1-Score, and MCC of 0**. With the lowest **AUC Score (0.577)** among all models, it demonstrated poor classification performance on this dataset. |
| **Random Forest** | Delivered the **best overall performance** with **93.66% Accuracy**, **0.862 AUC Score**, **84.62% Precision**, **50% Recall**, **62.86% F1-Score**, and **0.621 MCC**. It effectively balanced predictive performance and minority class detection, making it the most suitable model for deployment. |

## Overall feedback

> **Conclusion:** The dataset appears to be imbalanced, as some models achieved high accuracy while failing to predict any positive instances. Therefore, metrics such as **AUC Score, Recall, F1-Score, and MCC** provide a more reliable assessment than accuracy alone. Among all evaluated models, **Random Forest** demonstrated the best overall performance and is recommended as the final model.
